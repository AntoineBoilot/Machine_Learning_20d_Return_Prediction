```python
from pathlib import Path
import shutil
import subprocess
from datetime import datetime

REPO_PATH = Path.home() / "Desktop" / "Python" / "Machine_Learning_20d_Return_Prediction"
SOURCE_FOLDER = Path.home() / "Desktop" / "Python" / "Machine Learning Stocks" / "Fichier HTML"
DEST_FOLDER = REPO_PATH / "results"

FILES_TO_EXPORT = [
    "spx_dashboard.html",
    "cac40_dashboard.html",
    "dax_dashboard.html",
]

if not REPO_PATH.exists():
    raise FileNotFoundError(f"Repo not found: {REPO_PATH}")

if not SOURCE_FOLDER.exists():
    raise FileNotFoundError(f"Source folder not found: {SOURCE_FOLDER}")

if not DEST_FOLDER.exists():
    raise FileNotFoundError(f"Destination folder not found: {DEST_FOLDER}")


def run_git_command(cmd, cwd):
    result = subprocess.run(
        cmd,
        cwd=cwd,
        text=True,
        capture_output=True
    )
    print(f"$ {' '.join(cmd)}")
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed: {' '.join(cmd)}")
    return result.stdout.strip()


# 1) Vérifier que les fichiers source existent
for filename in FILES_TO_EXPORT:
    source_file = SOURCE_FOLDER / filename
    if not source_file.exists():
        raise FileNotFoundError(f"Missing source file: {source_file}")
    print(f"Found source: {source_file}")

# 2) Copier les fichiers vers results
for filename in FILES_TO_EXPORT:
    source_file = SOURCE_FOLDER / filename
    dest_file = DEST_FOLDER / filename
    shutil.copy2(source_file, dest_file)
    print(f"Copied: {source_file} -> {dest_file}")

# 3) Git add
run_git_command(
    ["git", "add"] + [str(DEST_FOLDER / f) for f in FILES_TO_EXPORT],
    cwd=REPO_PATH
)

# 4) Commit
timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
run_git_command(
    ["git", "commit", "--allow-empty", "-m", f"Refresh dashboard HTML files ({timestamp})"],
    cwd=REPO_PATH
)

# 5) Push sur main
run_git_command(["git", "push", "origin", "main"], cwd=REPO_PATH)

print("Done.")
print("Links:")
print("https://antoineboilot.github.io/Machine_Learning_20d_Return_Prediction/results/spx_dashboard.html")
print("https://antoineboilot.github.io/Machine_Learning_20d_Return_Prediction/results/cac40_dashboard.html")
print("https://antoineboilot.github.io/Machine_Learning_20d_Return_Prediction/results/dax_dashboard.html")
```
